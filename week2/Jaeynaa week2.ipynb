```python

#Q1)
def find_nash_equilibrium(matrix):

    m=len(matrix)
    n=len(matrix[0])

    
    p1_best = set()
    p2_best = set()

#a)
    for i in range(n):

        max_payoff=-999999

        for j in range(m):
            payoff = matrix[j][i][0]

            if payoff > max_payoff:

                max_payoff = payoff

            

        
        for r in range(m):

            if matrix[r][i][0]==max_payoff:

                p1_best.add((matrix[r][i][0],matrix[r][i][1]))

    
#b)
    
    for i in range(m):

        
        max_payoff=-999999

        for c in range(n):

            payoff = matrix[i][c][1]

            if payoff > max_payoff:

                max_payoff = payoff
        
        for j in range(n):

            if matrix[i][j][1]==max_payoff:

                p2_best.add((matrix[i][j][0],matrix[i][j][1]))

#c)

    nash=[]

    for cell in p1_best:

        if cell in p2_best:

            nash.append(cell)

#d)
    if len(nash)==0:

        print("No Pure Strategy Nash Equilibrium found")

    else:

        print("Pure Strategy Nash Equilibria:")

        for eq in nash:

            print(eq)

matrix = [
    [(4, 2), (0, 3)],   # U
    [(1, 4), (3, 0)]    # D
]
find_nash_equilibrium(matrix)

#b)p=4/5=0.8
#c)q=1/2=0.5
#d)((0.8,0.2),(0.5,0.5))
```

```python
#Q2)

    # General 2x2 game:
#
#           L        R
# U      (a,b)    (c,d)
# D      (e,f)    (g,h)
#
# p* = (h-f)/(b-f-d+h)
# q* = (g-c)/(a-c-e+g)

def mixed_ne(matrix):

    a,b = matrix[0][0]
    c,d = matrix[0][1]
    e,f = matrix[1][0]
    g,h = matrix[1][1]

    denom_p = b - f - d + h
    denom_q = a - c - e + g

    if denom_p == 0 or denom_q == 0:
        print("No interior mixed Nash Equilibrium exists.")
        return None

    p = (h - f) / denom_p
    q = (g - c) / denom_q

    if not (0 <= p <= 1 and 0 <= q <= 1):
        print("No valid interior mixed Nash Equilibrium.")
        return None

    print(f"p* = {p:.4f}")
    print(f"q* = {q:.4f}")

    return 


matching_pennies = [
    [(1,-1), (-1,1)],
    [(-1,1), (1,-1)]
]

mixed_ne(matching_pennies)

q1_game = [
    [(4,2), (0,3)],
    [(1,4), (3,0)]
]

mixed_ne(q1_game)

chicken = [
    [(0,0), (-1,1)],
    [(1,-1), (-10,-10)]
]

mixed_ne(chicken)
```

```python
#Q4)
import matplotlib.pyplot as plt
import numpy as np

# Prisoner's Dilemma payoffs
payoffs = {
    ('C', 'C'): (3, 3),
    ('C', 'D'): (0, 5),
    ('D', 'C'): (5, 0),
    ('D', 'D'): (1, 1)
}

# Part (a): Backward Induction
def backward_induction_pd(T):

    p1_total = 0
    p2_total = 0

    print(f"\nBackward Induction: T = {T}")

    for round_num in range(1, T + 1):

        # Backward induction predicts D in every round
        p1_action = 'D'
        p2_action = 'D'

        p1_payoff, p2_payoff = payoffs[(p1_action, p2_action)]

        p1_total += p1_payoff
        p2_total += p2_payoff

        print(
            f"Round {round_num}: "
            f"P1={p1_action}, P2={p2_action}, "
            f"Payoffs=({p1_payoff},{p2_payoff})"
        )

    print("Total Payoff P1 =", p1_total)
    print("Total Payoff P2 =", p2_total)

    return p1_total, p2_total


# Part (c): Tit-for-Tat vs Tit-for-Tat
def tit_for_tat_pd(T):

    p1_total = 0
    p2_total = 0

    print(f"\nTit-for-Tat vs Tit-for-Tat: T = {T}")

    prev_p1 = 'C'
    prev_p2 = 'C'

    for round_num in range(1, T + 1):

        if round_num == 1:
            p1_action = 'C'
            p2_action = 'C'
        else:
            p1_action = prev_p2
            p2_action = prev_p1

        p1_payoff, p2_payoff = payoffs[(p1_action, p2_action)]

        p1_total += p1_payoff
        p2_total += p2_payoff

        prev_p1 = p1_action
        prev_p2 = p2_action

        print(
            f"Round {round_num}: "
            f"P1={p1_action}, P2={p2_action}, "
            f"Payoffs=({p1_payoff},{p2_payoff})"
        )

    print("Total Payoff P1 =", p1_total)
    print("Total Payoff P2 =", p2_total)

    return p1_total, p2_total


# Part (b)
horizons = [5, 20, 100]

bi_totals = []
tft_totals = []

for T in horizons:

    bi_p1, bi_p2 = backward_induction_pd(T)
    tft_p1, tft_p2 = tit_for_tat_pd(T)

    bi_totals.append(bi_p1)
    tft_totals.append(tft_p1)


# Part (c): Bar Chart
x = np.arange(len(horizons))
width = 0.35

plt.figure(figsize=(8,5))

plt.bar(
    x - width/2,
    bi_totals,
    width,
    label="Backward Induction"
)

plt.bar(
    x + width/2,
    tft_totals,
    width,
    label="Tit-for-Tat"
)

plt.xticks(x, ['T=5', 'T=20', 'T=100'])

plt.ylabel("Total Payoff")
plt.xlabel("Time Horizon")
plt.title("Backward Induction vs Tit-for-Tat")
plt.legend()

plt.show()
```

```python
#Q5)
import numpy as np
import matplotlib.pyplot as plt

# Part (a)

def pv_stream(v, delta, T):

    total = 0

    for t in range(T):
        total += (delta ** t) * v

    return total


def pv_infinite(v, delta):

    return v / (1 - delta)


v = 6
delta = 0.9
T = 50

finite_pv = pv_stream(v, delta, T)
infinite_pv = pv_infinite(v, delta)

print("Finite PV =", finite_pv)
print("Infinite PV =", infinite_pv)


# Part (b)

def critical_delta(R, T_payoff, P):

    return (T_payoff - R) / (T_payoff - P)


# Part (c)

R = 3
T_payoff = 5
P = 1

delta_star = critical_delta(R, T_payoff, P)

print("Critical delta =", delta_star)

deltas = np.linspace(0.01, 0.99, 500)

pv_cooperate = []
pv_defect = []

for delta in deltas:

    # cooperate forever
    coop = R / (1 - delta)

    # defect once then receive punishment forever
    defect = T_payoff + delta * P / (1 - delta)

    pv_cooperate.append(coop)
    pv_defect.append(defect)

plt.figure(figsize=(10,6))

plt.plot(deltas, pv_cooperate, label="PV Cooperate")
plt.plot(deltas, pv_defect, label="PV Defect")

plt.axvline(
    x=delta_star,
    linestyle="--",
    label=f"delta* = {delta_star:.2f}"
)

plt.xlabel("Discount Factor (δ)")
plt.ylabel("Present Value")
plt.title("Cooperation vs Defection under Grim Trigger")
plt.legend()

plt.show()
```

```python
#Q6)
import matplotlib.pyplot as plt

# Stage game payoff matrix

payoffs = {
    ('H','H'):(6,6),
    ('H','L'):(0,9),
    ('L','H'):(9,0),
    ('L','L'):(2,2)
}




def always_defect(my_history, opp_history):
    return 'L'


def tit_for_tat(my_history, opp_history):

    if len(opp_history) == 0:
        return 'H'

    return opp_history[-1]


def grim_trigger(my_history, opp_history):

    if 'L' in opp_history:
        return 'L'

    return 'H'



def simulate_game(strategy1, strategy2, T):

    history1 = []
    history2 = []

    p1_total = 0
    p2_total = 0

    p1_cumulative = []
    p2_cumulative = []

    for _ in range(T):

        move1 = strategy1(history1, history2)
        move2 = strategy2(history2, history1)

        history1.append(move1)
        history2.append(move2)

        payoff1, payoff2 = payoffs[(move1, move2)]

        p1_total += payoff1
        p2_total += payoff2

        p1_cumulative.append(p1_total)
        p2_cumulative.append(p2_total)

    return (
        history1,
        history2,
        p1_total,
        p2_total,
        p1_cumulative,
        p2_cumulative
    )

#d)

T = 100

matchups = [
    ("Always Defect vs Tit-for-Tat",
     always_defect,
     tit_for_tat),

    ("Always Defect vs Grim Trigger",
     always_defect,
     grim_trigger),

    ("Tit-for-Tat vs Grim Trigger",
     tit_for_tat,
     grim_trigger)
]

results = []

for name, s1, s2 in matchups:

    result = simulate_game(s1, s2, T)

    results.append((name, result))

    print("\n", name)

    print("Firm 1 Total =", result[2])
    print("Firm 2 Total =", result[3])

#e)

fig, axes = plt.subplots(3, 1, figsize=(10, 12))

for ax, (name, result) in zip(axes, results):

    p1_cumulative = result[4]
    p2_cumulative = result[5]

    ax.plot(p1_cumulative, label="Firm 1")
    ax.plot(p2_cumulative, label="Firm 2")

    ax.set_title(name)
    ax.set_xlabel("Round")
    ax.set_ylabel("Cumulative Payoff")

    ax.legend()

plt.tight_layout()
plt.show()
```
