```python
#Q5)
import pandas as pd


data = {
    'Ticker': ['RELIANCE', 'TCS', 'INFY', 'HDFCBANK', 'WIPRO'],
    'Price': [2850.5, 3900.0, 1500.75, 1650.0, 420.3],
    'Volume': [1200000, 850000, 1100000, 950000, 700000],
    'Sector': ['Energy', 'IT', 'IT', 'Finance', 'IT']
}


df = pd.DataFrame(data)


df['Market_Cap_Proxy'] = df['Price'] * df['Volume']


#a) mean price 
mean_price=df['Price'].mean()
print("Mean Price =",mean_price)

#b)total volume traded in the 'IT' sector
tvol=df[df['Sector']=='IT']['Volume'].sum()
print("\ntotal volume traded in the 'IT' sector:",tvol)


#c)highs
highest_stock = df.loc[df['Market_Cap_Proxy'].idxmax(), 'Ticker']
print("Highest Market_Cap_Proxy Stock =", highest_stock)
```

```python
#Q6)
import numpy as np
scores = np.array([45, 82, 76, 55, 90, 38, 67, 71])

#a) Mean, Median, Standard Deviation
mean_score = np.mean(scores)
median_score = np.median(scores)
stddev = np.std(scores)
print("Mean =", mean_score)
print("Median =", median_score)
print("Standard Deviation =", stddev)

#b)Normalize(btw 0&1)
normalized_scores = (scores - np.min(scores)) / (np.max(scores) - np.min(scores))
print("\nnormalized scores:",normalized_scores)

#c)Indices of students scoring above mean
above_mean_indices=np.where(scores > mean_score)

print("\nIndices of students scoring above mean:")
print(above_mean_indices)

```

```python
#Q7)
import numpy as np
import matplotlib.pyplot as plt

#252 daily returns
daily_returns=np.random.normal(loc=0.0008,scale=0.015,size=252)

#a)price series starting at 100
price_series=100*np.cumprod(1+daily_returns)

#b)price series graph
plt.figure(figsize=(10, 5))
plt.plot(price_series)

plt.title("Simulated Stock Price Series")
plt.xlabel("Trading Days")
plt.ylabel("Price")

plt.grid(True)
plt.show()

#c.1)final price
final_price=price_series[-1]

#c.2)peak price
peak_price=np.max(price_series)

#c.3)Maximum drawdown
running_max=np.maximum.accumulate(price_series)
drawdowns=(price_series - running_max) / running_max
max_drawdown=np.min(drawdowns)

print(f"Final Price: {final_price:.2f}")
print(f"Peak Price: {peak_price:.2f}")
print(f"Maximum Drawdown: {max_drawdown:.2%}")



```

```python
#Q8)
import matplotlib.pyplot as plt
import numpy as np


months=['Jan','Feb','Mar','Apr','May','Jun',
          'Jul','Aug','Sep','Oct','Nov','Dec']

year1=[42, 45, 50, 48, 55, 60, 58, 63, 67, 70, 75, 80]

year2=[52, 49, 58, 61, 66, 71, 68, 74, 79, 83, 88, 95]

difference=np.array(year2)-np.array(year1)

colors=['green' if diff > 0 else 'red' for diff in difference]

fig,(ax1, ax2)=plt.subplots(2, 1, figsize=(12, 10))

# (a) Line chart

ax1.plot(months, year1,  marker='o', label='Year 1 Revenue')
ax1.plot(months, year2, marker='o', label='Year 2 Revenue')

ax1.set_title("Monthly Revenue Comparison")
ax1.set_xlabel("Months")
ax1.set_ylabel("Revenue (Lakhs)")

ax1.legend()
ax1.grid(True)


#b) Bar Chart
ax2.bar(months, difference, color=colors)

ax2.set_title("Revenue Difference (Year 2 - Year 1)")
ax2.set_xlabel("Months")
ax2.set_ylabel("Difference (Lakhs)")

ax2.axhline(0, color='black', linewidth=1)

ax2.grid(True)



plt.show()


```

```python
#Q9)
import numpy as np
import pandas as pd

#dataset 
df = pd.DataFrame({
    'Day': range(1, 11),

    'Price': [100, 103, np.nan, 108, 107,
              np.nan, 112, 115, np.nan, 120],

    'Volume': [500, 520, 510, np.nan, 530,
               545, np.nan, 560, 570, 580]
})



#a)missing values
print(df.isnull().sum())

#b)filling missing values
df['Price'] = df['Price'].ffill()

df['Volume'] = df['Volume'].fillna(df['Volume'].mean())

#c)Return column
df['Return'] = ((df['Price'] - df['Price'].shift(1))/df['Price'].shift(1))*100


print(df)
```

```python
#Q10)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


np.random.seed(42)

returns = np.random.multivariate_normal(
    mean=[0.0005, 0.0008, 0.0003],

    cov=[
        [0.0004, 0.0001, 0.00005],
        [0.0001, 0.0006, 0.00008],
        [0.00005, 0.00008, 0.0003]
    ],

    size=200
)

prices = 100 * np.cumprod(1 + returns, axis=0)


df = pd.DataFrame(
    prices,
    columns=['Asset_A', 'Asset_B', 'Asset_C']
)


print("Price Data:\n")
print(df.head())


plt.figure(figsize=(12, 6))

plt.plot(df['Asset_A'], label='Asset_A')
plt.plot(df['Asset_B'], label='Asset_B')
plt.plot(df['Asset_C'], label='Asset_C')

plt.title("Asset Price Series")
plt.xlabel("Trading Days")
plt.ylabel("Price")

plt.legend()
plt.grid(True)

daily_returns = df.pct_change()

correlation_matrix = daily_returns.corr()

print("\nCorrelation Matrix:\n")
print(correlation_matrix)

portfolio = (
    df['Asset_A'] +
    df['Asset_B'] +
    df['Asset_C']
) / 3


plt.plot(
    portfolio,
    label='Equal Weight Portfolio',
    linewidth=3
)

# Show final graph
plt.legend()
plt.show()
```
