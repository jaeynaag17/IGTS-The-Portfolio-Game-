```python
#Q2)
matrix = [
[(6, 4), (8, 1)], # P1 plays Row 0
[(4, 6), (7, 3)] # P1 plays Row 1
]
#a)
def dominant_strat_player1(matrix):
    
    num_rows=len(matrix)
    num_cols=len(matrix[0])
    dominant_row=None
    for i in range(num_rows):
        dominant=True
        for j in range(num_rows):
            if i!=j:
                for k in range(num_cols):
                    if matrix[i][k][0]<=matrix[j][k][0]:
                        dominant=False
                        break
        if dominant:
            dominant_row=i
            break

    if dominant_row is not None:
        print("Player 1 dominant strategy: Row", dominant_row)
    else:
        print("Player 1 has no dominant strategy")

dominant_strat_player1(matrix)

#b)
def dominant_strat_player2(matrix):
    
    num_rows=len(matrix)
    num_cols=len(matrix[0])
    dominant_cols=None
    for i in range(num_cols):
        dominant=True
        for j in range(num_cols):
            if i!=j:
                for k in range(num_rows):
                    if matrix[k][i][1]<=matrix[k][j][1]:
                        dominant=False
                        break
        if dominant:
            dominant_cols=i
            break

    if dominant_cols is not None:
        print("Player 2 dominant strategy: cols", dominant_cols)
    else:
        print("Player 2 has no dominant strategy")

dominant_strat_player2(matrix)


    
```

```python
#Q3)
def dominated_row(matrix):

    m=len(matrix)
    n=len(matrix[0])

    for i in range(m):

        for j in range(m):

            if i!=j:

                dominated=True

                for k in range(n):

                    
                    if matrix[i][k][0]>=matrix[j][k][0]:

                        dominated=False
                        break

                if dominated:
                    return i,j

    return None


def dominated_col(matrix):

    m=len(matrix)
    n=len(matrix[0])

    for i in range(n):

        for j in range(n):

            if i!=j:

                dominated=True

                for k in range(m):

                    # Compare Player 2 payoffs
                    if matrix[k][i][1] >= matrix[k][j][1]:

                        dominated=False
                        break

                if dominated:
                    return i,j

    return None


def remove_row(matrix, row_index):

    matrix.pop(row_index)


def remove_col(matrix, col_index):

    for row in matrix:

        row.pop(col_index)


def iesds(matrix):

    while True:

        changes=False

       
        row_result=dominated_row(matrix)

        if row_result is not None:

            dominated_r, dominating_r=row_result

            print("Removing Row", dominated_r,
                  "because Row", dominating_r,
                  "strictly dominates it")

            remove_row(matrix, dominated_r)

            changes=True

        
        col_result=dominated_col(matrix)

        if col_result is not None:

            dominated_c, dominating_c=col_result

            print("Removing Column", dominated_c,
                  "because Column", dominating_c,
                  "strictly dominates it")

            remove_col(matrix, dominated_c)

            changes=True

        if not changes:
            break

    return matrix



matrix=[

    [(1,1), (-1,2), (5,0), (1,1)],

    [(2,3), (1,2), (3,0), (5,1)],

    [(1,1), (0,5), (1,7), (0,1)]

]

final_matrix = iesds(matrix)

print("\nFinal Remaining Matrix:")

for row in final_matrix:
    print(row)

if len(final_matrix) == 1 and len(final_matrix[0]) == 1:

    print("\nPredicted Outcome:", final_matrix[0][0])
```

```python

#Q4)
def find_nash_equilibrium(matrix):

    m=len(matrix)
    n=len(matrix[0])

    
    p1_best = set()
    p2_best = set()

#a)
    for i in range(n):

        max_payoff=-999999

        for j in range(m):
            payoff = matrix[j][i][0]

            if payoff > max_payoff:

                max_payoff = payoff

            

        
        for r in range(m):

            if matrix[r][i][0]==max_payoff:

                p1_best.add((matrix[r][i][0],matrix[r][i][1]))

    
#b)
    
    for i in range(m):

        
        max_payoff=-999999

        for c in range(n):

            payoff = matrix[i][c][1]

            if payoff > max_payoff:

                max_payoff = payoff
        
        for j in range(n):

            if matrix[i][j][1]==max_payoff:

                p2_best.add((matrix[i][j][0],matrix[i][j][1]))

#c)

    nash=[]

    for cell in p1_best:

        if cell in p2_best:

            nash.append(cell)

#d)
    if len(nash)==0:

        print("No Pure Strategy Nash Equilibrium found")

    else:

        print("Pure Strategy Nash Equilibria:")

        for eq in nash:

            print(eq)

#1)
Prisoner_Dilemma=[

    [(-1,-1), (-3,0)],

    [(0,-3), (-2,-2)]

]
find_nash_equilibrium(Prisoner_Dilemma)
#2)
stag_hare=[

    [(4,4), (0,2)],

    [(2,0), (2,2)]

]
find_nash_equilibrium(stag_hare)

#3)
chicken=[

    [(0,0), (-1,1)],

    [(1,-1), (-10,-10)]

]
find_nash_equilibrium(chicken)
#4)
bos=[

    [(2,1), (0,0)],

    [(0,0), (1,2)]

]
find_nash_equilibrium(bos)

m=[

    [(4,4), (0,0), (1,3)],

    [(0,0), (5,5), (0,0)],

    [(3,1), (0,0), (6,6)]

]
find_nash_equilibrium(m)
```

```python
#Q50
def generate_profiles(strategies, current_profile, player, all_profiles):


    if player==len(strategies):

        all_profiles.append(tuple(current_profile))
        return

   
    for strategy in strategies[player]:

        current_profile.append(strategy)

        generate_profiles(strategies,
                          current_profile,
                          player + 1,
                          all_profiles)

        current_profile.pop()


def find_nash_n_player(strategies, payoffs):

    n_players = len(strategies)

    
    all_profiles = []

    generate_profiles(strategies, [], 0, all_profiles)

    nash_equilibria = []

    

    for profile in all_profiles:

        is_nash = True

        
        for player in range(n_players):

            current_payoff = payoffs[profile][player]

           
            for new_strategy in strategies[player]:

                
                if new_strategy == profile[player]:

                    continue

                
                new_profile = list(profile)

                new_profile[player] = new_strategy

                new_profile = tuple(new_profile)

                new_payoff = payoffs[new_profile][player]

                
                if new_payoff > current_payoff:

                    is_nash = False
                    break

            if not is_nash:
                break

        
        if is_nash:

            nash_equilibria.append(profile)

    

    if len(nash_equilibria) == 0:

        print("No Pure Strategy Nash Equilibrium")

    else:

        print("Pure Strategy Nash Equilibria:")

        for eq in nash_equilibria:

            print(eq)




strategies = [

    ['C', 'D'],
    ['C', 'D'],
    ['C', 'D']

]

payoffs = {

    ('C','C','C'): [4,4,4],

    ('C','C','D'): [1,1,6],

    ('C','D','C'): [1,6,1],

    ('C','D','D'): [0,3,3],

    ('D','C','C'): [6,1,1],

    ('D','C','D'): [3,0,3],

    ('D','D','C'): [3,3,0],

    ('D','D','D'): [2,2,2]

}

find_nash_n_player(strategies, payoffs)
#If one player changes their decision while the others keep theirs the same, they do not get a better result at ('D','D','D').

#So all players prefer to stay with their current choice, making it a Nash Equilibrium.
```

```python
#Q6)
def find_dominant_strategy(matrix):

    m = len(matrix)
    n = len(matrix[0])

    p1_dominant = None
    p2_dominant = None

   
    for r1 in range(m):

        dominant = True

        for r2 in range(m):

            if r1 != r2:

                for c in range(n):

                    if matrix[r1][c][0] <= matrix[r2][c][0]:

                        dominant = False
                        break

        if dominant:

            p1_dominant = r1
            break

    
    for c1 in range(n):

        dominant = True

        for c2 in range(n):

            if c1 != c2:

                for r in range(m):

                    if matrix[r][c1][1] <= matrix[r][c2][1]:

                        dominant = False
                        break

        if dominant:

            p2_dominant = c1
            break

    
    print("\n--- Dominant Strategy Analysis ---")

    if p1_dominant is not None:

        print("Player 1 dominant strategy: Row", p1_dominant)

    else:

        print("Player 1 has no dominant strategy")

    if p2_dominant is not None:

        print("Player 2 dominant strategy: Column", p2_dominant)

    else:

        print("Player 2 has no dominant strategy")



def find_dominated_row(matrix):

    m = len(matrix)
    n = len(matrix[0])

    for r1 in range(m):

        for r2 in range(m):

            if r1 != r2:

                dominated = True

                for c in range(n):

                    if matrix[r1][c][0] >= matrix[r2][c][0]:

                        dominated = False
                        break

                if dominated:

                    return r1, r2

    return None


def find_dominated_col(matrix):

    m = len(matrix)
    n = len(matrix[0])

    for c1 in range(n):

        for c2 in range(n):

            if c1 != c2:

                dominated = True

                for r in range(m):

                    if matrix[r][c1][1] >= matrix[r][c2][1]:

                        dominated = False
                        break

                if dominated:

                    return c1, c2

    return None


def remove_row(matrix, row_index):

    matrix.pop(row_index)


def remove_col(matrix, col_index):

    for row in matrix:

        row.pop(col_index)


def iesds(matrix):

    print("\n--- IESDS Analysis ---")

    while True:

        changed = False

        row_result = find_dominated_row(matrix)

        if row_result is not None:

            dominated_row, dominating_row = row_result

            print("Removing Row", dominated_row,
                  "because Row", dominating_row,
                  "strictly dominates it")

            remove_row(matrix, dominated_row)

            changed = True

        col_result = find_dominated_col(matrix)

        if col_result is not None:

            dominated_col, dominating_col = col_result

            print("Removing Column", dominated_col,
                  "because Column", dominating_col,
                  "strictly dominates it")

            remove_col(matrix, dominated_col)

            changed = True

        if not changed:

            break

    print("\nRemaining Matrix:")

    for row in matrix:

        print(row)



def find_nash_equilibrium(matrix):

    m = len(matrix)
    n = len(matrix[0])

    p1_best = set()
    p2_best = set()

    # P1 best responses
    for c in range(n):

        max_payoff = -999999

        for r in range(m):

            payoff = matrix[r][c][0]

            if payoff > max_payoff:

                max_payoff = payoff

        for r in range(m):

            if matrix[r][c][0] == max_payoff:

                p1_best.add((r, c))

    
    for r in range(m):

        max_payoff = -999999

        for c in range(n):

            payoff = matrix[r][c][1]

            if payoff > max_payoff:

                max_payoff = payoff

        for c in range(n):

            if matrix[r][c][1] == max_payoff:

                p2_best.add((r, c))

    
    nash = []

    for cell in p1_best:

        if cell in p2_best:

            nash.append(cell)

    
    print("\n--- Nash Equilibrium Analysis ---")

    if len(nash) == 0:

        print("No Pure Strategy Nash Equilibrium")

    else:

        print("Pure Strategy Nash Equilibria:")

        for eq in nash:

            print(eq)



m = int(input("Enter number of rows (P1 strategies): "))
n = int(input("Enter number of columns (P2 strategies): "))

matrix = []

print("\nEnter payoffs row by row")

for i in range(m):

    row = []

    for j in range(n):

        print("\nCell [", i, ",", j, "]")

        p1 = int(input("Enter P1 payoff: "))
        p2 = int(input("Enter P2 payoff: "))

        row.append((p1, p2))

    matrix.append(row)


print("\n--- Entered Matrix ---")

for row in matrix:

    print(row)


find_dominant_strategy(matrix)

copy_matrix = []

for row in matrix:

    copy_matrix.append(row[:])

iesds(copy_matrix)

find_nash_equilibrium(matrix)
```
